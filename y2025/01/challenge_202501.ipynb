```python
def lines_split(input_string):
  lines = input_string.strip().split('\n')
  return [line.strip() for line in lines]
```

```python
def read_lines_from_file(file_path):
  with open(file_path, 'r') as f_in:
      return lines_split(f_in.read())
```

```python
def convert_directions_to_integers(directions):
  result = [num for num in map(lambda x: int(x.replace('L', '-').replace('R', '')), directions)]
  return result
```

```python
def challenge_202501_part1(rotations, start_position=0):
  zero_count = 0
  dial = start_position
  positions = []
  for rotation in rotations:
    result = dial + rotation
    dial = result % 100
    if dial == 0:
      zero_count += 1
    positions.append(dial)
  return {'zero':zero_count, 'positions': positions, 'final_position': dial}
```

```python
directions = read_lines_from_file('./input.txt')
rotations = convert_directions_to_integers(directions)
results1 = challenge_202501_part1(rotations, 50)
results1['zero']
```

```python
def challenge_202501_part2(rotations, start_position=0):
  zero_count = 0
  dial = start_position
  positions = []
  for rotation in rotations:
    start = dial
    result = dial + rotation
    dial = result % 100
    if dial == 0:
      zero_count += 1
    positions.append((start, rotation, dial, zero_count))
  return {'zero':zero_count, 'positions': positions, 'final_position': dial}
```

```python
results2 = challenge_202501_part2(rotations, 50)
results2['zero']
```
