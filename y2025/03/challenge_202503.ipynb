## Parsing, Solution Functions, and Input Reader

This section contains:

- `read_input_file(path)` → reads the real puzzle input from a `.txt` file  
- `parse_input(raw)` → converts a string into a useful structure  
- `solve_part1(raw)` and `solve_part2(raw)` → implementations for each part  


```python
def read_input_file(path: str) -> str:
    """
    Reads a .txt file and returns its content as a string.
    Used for real puzzle input, not for tests.
    """
    with open(path, "r", encoding="utf-8") as f:
        return f.read().strip()


def parse_input(raw: str):
    """
    Parses the raw puzzle input into whatever structure is needed.
    Modify this function according to each puzzle.
    """
    data = raw.splitlines()
    return [list(char for char in line) for line in data] 

def get_line_joltage(line):
    index_start = line.index(max(line[:-1]))
    index_end = line.index(max(line[index_start+1:]))
    return line[index_start]+line[index_end]
    
def get_line_joltage_nth(line, n):
    remainder = n
    joltage = ""
    last_index = -1
    while remainder > 0:
        print(remainder)
        new_last_index = line.index(max(line[last_index+1:-remainder]))
        new_char = line[new_last_index]
        joltage += new_char
        remainder -= 1
        last_index = new_last_index
        print(joltage)
    return joltage
    

def solve_part1(raw: str):
    """Solves Part 1 of the puzzle."""
    data = parse_input(raw)
    return sum(int(get_line_joltage(line)) for line in data)


def solve_part2(raw: str):
    """Solves Part 2 of the puzzle."""
    data = parse_input(raw)
    return sum(int(get_line_joltage_nth(line,12)) for line in data)
```
