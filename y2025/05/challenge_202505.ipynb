## Parsing, Solution Functions, and Input Reader

This section contains:

- `read_input_file(path)` → reads the real puzzle input from a `.txt` file  
- `parse_input(raw)` → converts a string into a useful structure  
- `solve_part1(raw)` and `solve_part2(raw)` → implementations for each part  


```python
def read_input_file(path: str) -> str:
    """
    Reads a .txt file and returns its content as a string.
    Used for real puzzle input, not for tests.
    """
    with open(path, "r", encoding="utf-8") as f:
        return f.read().strip()


def parse_input(raw: str):
    """
    Parses the raw puzzle input into whatever structure is needed.
    Modify this function according to each puzzle.
    """
    ranges, numbers = raw.split("\n\n") 
    range_lines = []
    for line in ranges.splitlines():
        start, end = line.split("-")
        range_lines.append((int(start), int(end)))
    nember_lines = [int(line) for line in numbers.splitlines()]
    return range_lines, nember_lines

def get_ids_in_range(rng):
    """Returns a set of all IDs in the given range (inclusive)."""
    start, end = rng
    return list(range(start, end + 1))

def get_all_ids(ranges):
    """Returns a set of all IDs covered by the given ranges."""
    all_ids = set()
    for rng in ranges:
        ids_in_range = get_ids_in_range(rng)
        all_ids.update(ids_in_range)
    return all_ids

def count_unique_ids_in_ranges(ranges):
    """
    Dado una lista de rangos (start, end), devuelve
    cuántas IDs únicas hay en total, sin expandir a listas.
    """
    if not ranges:
        return 0

    # Ordenar por inicio
    sorted_ranges = sorted(ranges, key=lambda x: x[0])

    merged = []
    cur_start, cur_end = sorted_ranges[0]

    for start, end in sorted_ranges[1:]:
        # Si se solapan o se tocan (ej: [1,5] y [6,10])
        if start <= cur_end + 1:
            # extendemos el intervalo actual
            cur_end = max(cur_end, end)
        else:
            # cerramos el intervalo actual y empezamos uno nuevo
            merged.append((cur_start, cur_end))
            cur_start, cur_end = start, end

    # agregar el último
    merged.append((cur_start, cur_end))

    # sumar la longitud de cada intervalo fusionado
    total = sum(end - start + 1 for start, end in merged)
    return total

def solve_part1(raw: str):
    """Solves Part 1 of the puzzle."""
    data = parse_input(raw)
    ranges, numbers = data  
    count = 0
    for num in numbers:
        if any(start <= num <= end for start, end in ranges):
            count += 1
    return count


def solve_part2(raw: str) -> int:
    data = parse_input(raw)
    ranges, numbers = data  # o como lo tengas estructurado
    return count_unique_ids_in_ranges(ranges)

```
