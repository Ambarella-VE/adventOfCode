## Parsing, Solution Functions, and Input Reader

This section contains:

- `read_input_file(path)` → reads the real puzzle input from a `.txt` file  
- `parse_input(raw)` → converts a string into a useful structure  
- `solve_part1(raw)` and `solve_part2(raw)` → implementations for each part  


```python
def read_input_file(path: str) -> str:
    """
    Reads a .txt file and returns its content as a string.
    Used for real puzzle input, not for tests.
    """
    with open(path, "r", encoding="utf-8") as f:
        return f.read().strip()


def parse_input_1(raw: str):
    """
    Parses the raw puzzle input into whatever structure is needed.
    Modify this function according to each puzzle.
    """
    lines = raw.splitlines()
    data = [line.split(' ') for line in lines]
    cleaned_data = [[line for line in line if line != ''] for line in data]
    return cleaned_data

import numpy as np

def parse_input(raw: str):
    lines = raw.splitlines()
    ops = lines[-1]
    start_idx = []
    for idx, char in enumerate(ops):
        if char in ('+', '*'):
            start_idx.append(idx)
    ops = [char for char in ops if char in ('+', '*')]
    lines = lines[:-1]
    mat = np.matrix([list(line) for line in lines])
    transposed = mat.T.tolist()
    concat = [''.join(row) for row in transposed]
    nums = [int(num) if num.strip().isdigit() else 0 for num in concat]
    nums_lst = split_by_zero(nums)
    return nums_lst,ops

def split_by_zero(lst):
    chunks = []
    current = []

    for x in lst:
        if x == 0:
            if current:
                chunks.append(current)
                current = []
        else:
            current.append(x)

    if current:
        chunks.append(current)

    return chunks



def solve_part1(raw: str):
    """Solves Part 1 of the puzzle."""
    data = parse_input_1(raw)
    columns = data[:-1]
    operations = data[-1]
    total = 0
    for index, op in enumerate(operations):
        if op not in ('+', '*'):
            raise ValueError(f"Unsupported operation: {op}")
        else:
            if '+' == op:
                result = sum(int(columns[row][index]) for row in range(len(columns)))
                total += result
            if '*' == op:
                result = 1
                for row in range(len(columns)):
                    result *= int(columns[row][index])
                total += result
    return total


def solve_part2(raw: str):
    """Solves Part 2 of the puzzle."""
    nums, ops = parse_input(raw)
    total = 0
    for index, op in enumerate(ops):
        if op not in ('+', '*'):
            raise ValueError(f"Unsupported operation: {op}")
        else:
            if '+' == op:
                result = sum(nums[index])
                total += result
            if '*' == op:
                result = 1
                for num in nums[index]:
                    result *= num
                total += result
    return total
```
